# LitAudit
Audit literature-derived materials databases: check each value against another database and against physics relations; diagnose disagreements as a known unit/quantity factor or as unexplained (with the nearest factor shown).

```python
import pandas as pd
from loaders import load_all
from audit import cross_db, relations, planted_benchmark
pd.set_option('display.max_colwidth', 45)
df = load_all()
df.groupby(['db', 'property']).size()
```

## 1. Cross-database check (same paper + composition in two databases)

```python
xdb = cross_db(df)
print(xdb.diagnosis.value_counts().to_string())
xdb[xdb.diagnosis != 'agree'].sort_values('resid', key=abs, ascending=False)
```

## 2. Physics-relation check (hardness vs yield strength; measured vs calculated density)

```python
rel = relations(df)
print(rel.groupby(['check', 'diagnosis']).size().to_string())
rel[rel.diagnosis != 'agree'].sort_values('resid', key=abs, ascending=False)
```

## 3. Planted-error benchmark

```python
planted_benchmark(pd.concat([xdb, rel])).round(2)
```
