# LitAudit
Audit literature-derived materials databases: compare each value with independent references (another database, its own temperature series, other papers, physics relations) and diagnose disagreements as a named unit/quantity slip or as unexplained (nearest cause shown).

```python
import pandas as pd
from loaders import load_all
from audit import run, entry_verdicts, coverage, planted_benchmark
pd.set_option('display.max_colwidth', 45)
df = load_all()
checks = run(df)
checks.groupby(['check', 'diagnosis']).size()
```

## 1. Flags, largest gap first

```python
flags = checks[checks.diagnosis != 'agree'].sort_values('resid', key=abs, ascending=False)
flags[['check', 'composition', 'value', 'reference', 'resid', 'attributable', 'diagnosis', 'nearest', 'doi']].head(30)
```

## 2. Per-entry verdicts and coverage

```python
print(entry_verdicts(checks).verdict.value_counts())
coverage(df, checks)
```
